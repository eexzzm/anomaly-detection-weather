```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_parquet("../data/interim/features.parquet")
```

```python
# which features generate NaN

nan_summary = df.isna().mean().sort_values(ascending=False)
nan_summary
```

the highest is around 1-2 percent, which is a rare NaN not structural.

```python
row_nan_ratio = df.isna().mean(axis=1)
print(row_nan_ratio.describe())
print(f"percentage of rows that will be dropped: {(row_nan_ratio > 0).mean()}")
```

```python
plt.plot(row_nan_ratio.values)
plt.title("NaN density over time")
```
